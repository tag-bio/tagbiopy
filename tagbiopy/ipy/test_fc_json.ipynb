```python
test = {"fc": {
    "name": "fc-local",
    "url": "http://localhost:8000/",
    "api_key": ""
  }}
```

```python
from collections import namedtuple
```

```python
def fc_decoder(payload):
    return namedtuple('FC', payload.keys())(*payload.values())
```

```python
wtf = fc_decoder(test['fc'])
```
